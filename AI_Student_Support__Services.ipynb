# * Import Libraries

This section imports the essential Python libraries required for data processing, NLP, Machine Learning, model evaluation, and chatbot development.

These libraries help prepare the dataset, train the model, evaluate its performance, and create the interactive student support chatbot.

```python
# AI Student Support Services
# Step 1: Import Libraries

import pandas as pd
import numpy as np
import re

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

print("All libraries imported successfully!")
```

# • Load Dataset

This section loads the university-provided student query dataset into the Python environment. The dataset is then used for data analysis, preprocessing, and Machine Learning model development.

```python
# Step 2: Load Dataset

train_df = pd.read_csv("/content/university_query_train.csv")
test_df = pd.read_csv("/content/university_query_test.csv")

print("Training Dataset Shape:", train_df.shape)
print("Testing Dataset Shape:", test_df.shape)

display(train_df.head())
```

# • Data Understanding

This section explores the dataset structure, features, data types, and missing values to understand the data before applying preprocessing and Machine Learning techniques.

```python
# Step 3: Basic Data Understanding

print("Training Dataset Information")
print("=" * 40)

print("Rows:", train_df.shape[0])
print("Columns:", train_df.shape[1])

print("\nColumns:")
print(train_df.columns.tolist())

print("\nMissing Values:")
print(train_df.isnull().sum())

print("\nPriority Distribution:")
print(train_df["Priority_Label"].value_counts())
```

# • Text Cleaning

This section cleans and standardizes student queries by converting text to lowercase and removing unnecessary characters and spaces, making the data ready for NLP and Machine Learning.

```python
# Step 4: Text Preprocessing

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

train_df["Cleaned_Query"] = train_df["Student_Query"].apply(clean_text)
test_df["Cleaned_Query"] = test_df["Student_Query"].apply(clean_text)

print("Text preprocessing completed successfully!")

display(
    train_df[["Student_Query", "Cleaned_Query"]].head()
)
```

# • TF - IDF

TF-IDF converts the cleaned student queries into numerical features by identifying the importance of words. These features are then used by the Machine Learning model for classification.

```python
# Step 5: TF-IDF Vectorization

tfidf = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=3000
)

X_train = tfidf.fit_transform(train_df["Cleaned_Query"])
X_test = tfidf.transform(test_df["Cleaned_Query"])

y_train = train_df["Priority_Label"]
y_test = test_df["Priority_Label"]

print("TF-IDF completed successfully!")
print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)
```

# • Train Machine Learning Model

This section trains the Logistic Regression model using the TF-IDF features extracted from student queries. The model learns patterns from the training data and predicts the priority level of new student queries.

```python
# Step 6: Train Machine Learning Model

model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model.fit(X_train, y_train)

print("Machine Learning Model trained successfully!")
```

# • Evaluate Model

This section evaluates the performance of the trained Machine Learning model using accuracy and classification metrics such as precision, recall, and F1-score.

```python
# Step 7: Model Evaluation

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

## • AI Student Support Bot


AI Student Support Bot is an AI-based chatbot that helps students by answering common academic and support-related queries. It uses NLP, TF-IDF, and Machine Learning to understand user questions and provide relevant responses quickly and efficiently.

```python
# Step 8: AI Student Support Bot

def student_support_bot(question):

    # Clean user question
    cleaned_question = clean_text(question)

    # Convert question into TF-IDF
    question_vector = tfidf.transform([cleaned_question])

    # Predict category/priority
    prediction = model.predict(question_vector)[0]

    # Confidence
    probabilities = model.predict_proba(question_vector)[0]
    confidence = probabilities.max() * 100

    print("=" * 55)
    print("        AI STUDENT SUPPORT SERVICES")
    print("=" * 55)
    print("Student:", question)
    print("Prediction:", prediction)
    print("Confidence:", round(confidence, 2), "%")

    print("\nAI Support Bot:")

    if prediction == "High":
        print("Your query is marked as HIGH priority.")
        print("Please contact the concerned department.")

    elif prediction == "Medium":
        print("Your query is marked as MEDIUM priority.")
        print("The concerned department can assist you.")

    else:
        print("Your query is marked as LOW priority.")
        print("Please check the regular student support resources.")

    print("=" * 55)
```

## • Test the Bot


The chatbot is tested with different student queries to check whether it correctly identifies the intent and provides the relevant response. This helps evaluate the bot’s accuracy and reliability.

```python
student_support_bot(
    "I cannot download my hall ticket for tomorrow"
)
```

# • Actual Chatbot Interface 🤖

The final chatbot interface allows students to enter their questions and receive instant, relevant responses in a simple and user-friendly format.

```python
# Step 9: Interactive Chatbot Interface

!pip -q install gradio
```

```python
import gradio as gr

def chatbot_response(question):

    cleaned_question = clean_text(question)

    question_vector = tfidf.transform([cleaned_question])

    prediction = model.predict(question_vector)[0]

    probabilities = model.predict_proba(question_vector)[0]
    confidence = probabilities.max() * 100

    if prediction == "High":
        response = "Your query has HIGH priority. Please contact the concerned department."
    elif prediction == "Medium":
        response = "Your query has MEDIUM priority. The concerned department can assist you."
    else:
        response = "Your query has LOW priority. Please check regular support resources."

    return (
        f"🤖 **AI Student Support Services**\n\n"
        f"**Priority:** {prediction}\n\n"
        f"**Confidence:** {confidence:.2f}%\n\n"
        f"**Response:** {response}"
    )

demo = gr.Interface(
    fn=chatbot_response,
    inputs=gr.Textbox(
        label="Ask your question",
        placeholder="Type your student query here..."
    ),
    outputs=gr.Markdown(),
    title="🎓 AI Student Support Services",
    description="Ask a university-related question and get an AI-based priority prediction."
)

demo.launch()
```
